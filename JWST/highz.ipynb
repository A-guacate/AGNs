```python
import pandas as pd
import os

# --- Carga del catálogo ---
local_file = "JWST_AGN_catalog_v1.0.0.csv"
if os.path.exists(local_file):
    catalogue = pd.read_csv(local_file)
else:
    url = "https://github.com/VirginiaLenk1/AGNs-in-eJWST/releases/download/AGN_catalogue/JWST_AGN_catalog_v1.0.0.csv"
    catalogue = pd.read_csv(url)
    catalogue.to_csv(local_file, index=False)

# --- Filtrado por redshift alto (4 <= z <= 8) ---
mask_z = (catalogue["z"] >= 4) & (catalogue["z"] <= 8)
high_z = catalogue[mask_z].copy()

# --- Elimino duplicados
unique_agn = high_z.drop_duplicates(subset="Name")[
    ["Name", "RA (J2000, deg)", "Dec (J2000, deg)", "objType", "Ref",
     "z", "zType", "zRef", "Targeted"]
].copy()

obs_count = high_z.groupby("Name").size().rename("N_Observaciones")
unique_agn = unique_agn.merge(obs_count, on="Name")

unique_agn = unique_agn.sort_values("z").reset_index(drop=True)
unique_agn.to_csv("AGN_high_z_4_8.csv", index=False)

print(f"AGNs únicos con 4<=z<=8: {len(unique_agn)}")
print(unique_agn["Targeted"].value_counts())
```
